```python
q=df.iloc[:,3:].values
q
```

```python
from sklearn.cluster import KMeans
```

```python
cs=[]
for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(q)
    cs.append(kmeans.inertia_)
plt.plot(cs)
```

```python
kmeans=KMeans(n_clusters=5,init='k-means++',random_state=42)
y_kmeans=kmeans.fit_predict(q)
```

```python
kmeans.inertia_
```

```python
kmeans.labels_
```

```python
plt.scatter(q[y_kmeans==0,0],q[y_kmeans==0,1],color='green',label='cluster1')
plt.scatter(q[y_kmeans==1,0],q[y_kmeans==1,1],color='blue',label='cluster2')
plt.scatter(q[y_kmeans==2,0],q[y_kmeans==2,1],color='red',label='cluster3')
plt.scatter(q[y_kmeans==3,0],q[y_kmeans==3,1],color='yellow',label='cluster4')
plt.scatter(q[y_kmeans==4,0],q[y_kmeans==4,1],color='pink',label='cluster5')
plt.scatter(kmeans.cluster_centers_[:,0],kmeans.cluster_centers_[:,1],color='red',label='centroid',marker='>')
```
